# 00 - Setup Environment

Creates the Unity Catalog catalog and schemas used by the ABC Retail
Lakehouse. Run this notebook once before the remaining notebooks.

```python
import re

dbutils.widgets.text("catalog", "abc_retail", "Target catalog")
dbutils.widgets.text(
    "landing_root",
    "/Volumes/abc_retail/dev/landing",
    "Landing root",
)

catalog = dbutils.widgets.get("catalog").strip()
landing_root = dbutils.widgets.get("landing_root").strip().rstrip("/")

if not re.fullmatch(r"[A-Za-z_][A-Za-z0-9_]*", catalog):
    raise ValueError(f"Invalid catalog identifier: {catalog!r}")

if not landing_root:
    raise ValueError("landing_root must not be empty")
```

```python
spark.conf.set("spark.sql.session.timeZone", "UTC")

spark.sql(f"CREATE CATALOG IF NOT EXISTS `{catalog}`")
spark.sql(f"USE CATALOG `{catalog}`")

for schema_name in ["bronze", "silver", "gold", "quarantine", "control"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema_name}`")
```

```python
spark.sql(f"SHOW SCHEMAS IN `{catalog}`").display()

print(f"Catalog ready: {catalog}")
print(f"Landing root configured for later notebooks: {landing_root}")
```
